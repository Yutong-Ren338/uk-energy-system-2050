# Electrochemical DAC (E-DAC)

E-DAC ("electricity-driven DAC") is a form of direct air capture that pulls CO2 out of the atmosphere using electricity rather than heat. The real-world process, bipolar-membrane electrodialysis (BPMED), absorbs CO2 from air into an alkaline solution in a contactor, then regenerates that solution electrochemically: a bipolar membrane splits water into H+ and OH- under an applied voltage, the acid stream drops the solution's pH and strips the CO2 back out as a gas, and the regenerated base stream is recycled to the contactor. This matches the process modelled in Sabatino et al. (2020) *Ind. Eng. Chem. Res.* 59(15) and, with more numerical detail on the energy demand, in Liu, Yang & Darton (2024) *ACS Sustainable Chem. Eng.* 12(10) - whose ideal-case energy limits (~154-162 kJ/mol) bound the same range this model uses.

In this repository, that process is represented as a fixed energy intensity (a "Medium" scenario of 101 kJ per mol CO2, from a modelled low/medium/high range of 43-162 kJ/mol - `src/assumptions.py` `DAC.EnergyCost`) converting installed electrical power into a nameplate CO2 capture capacity, with an equipment-only cost model and Wright's-law learning curve on top (`src/assumptions.py` `EDAC`, `src/costs.py` `edac_equipment_costs`, reference plant per Leonzio & Shah (2025) *RSC Sustainability*). E-DAC only ever runs on surplus electricity, and only after medium-term storage and hydrogen storage (in the hydrogen-pathway variant) have already been filled; anything left over after E-DAC's installed capacity is curtailed (`src/power_system_core.py` `handle_surplus`/`handle_dac`).
