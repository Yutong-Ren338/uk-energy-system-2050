# Model Structure

This model is a **daily-resolution, rule-based dispatch simulation** for the UK power system (not a linear-programming optimizer). For a *fixed* technology mix, it steps through ~40+ years of historical weather-driven demand and renewable output, day by day, balancing supply and demand with a fixed priority order: on a deficit day it draws on interconnector imports, then storage, then gas generation with carbon capture, then (in the hydrogen-pathway variant) hydrogen storage; on a surplus day it charges storage, then runs direct air capture on the remainder, curtailing whatever is left.

Capacity **optimization happens outside this core simulation**: an outer scenario-search loop sweeps renewable and DAC capacities and bisection-searches for the cheapest mix that hits a target net CO2 removal, re-running the simulation at each trial. Costs are computed by a separate accounting layer that closes a feedback loop with the simulation itself: realised curtailment from a run is fed back in to compute a curtailment-adjusted cost, which then determines the final system cost.
